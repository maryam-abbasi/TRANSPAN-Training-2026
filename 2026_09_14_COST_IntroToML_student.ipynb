## Task 4 of 14 - Loops

### 1. Concept

A **loop** repeats the same instruction for every item in a list. This is why a
computer can apply one rule to 569 patients without anyone typing it 569 times.

Two details matter and both are about layout. The first line ends with a **colon
`:`**, and the repeated lines are **indented** by four spaces. In Python the
indentation is not decoration - it is what marks which lines belong inside the
loop.

### 2. Code Breakdown

Read it as: "for each item in the list `features`, call it `feature`, and do the
indented part".

The name `feature` is our choice; it is a temporary variable that takes each

of times rather than walk through a list:

Our objective with this dataset is to construct a machine learning model capable of utilizing tumor-related information to make predictions about the malignancy or benign nature of tumors.

### Variable Description
Attribute Information:

- ID number
- Diagnosis (M = malignant, B = benign)

Ten real-valued features are computed for each cell nucleus:

- Radius (mean of distances from center to points on the perimeter)
- Texture (standard deviation of gray-scale values)
- Perimeter
- Area
- Smoothness (local variation in radius lengths)
- Compactness (perimeter^2 / area - 1.0)
- Concavity (severity of concave portions of the contour)
- Concave points (number of concave portions of the contour)
- Symmetry
- Fractal dimension ("coastline approximation" - 1)

The mean, standard error and "worst" or largest (mean of the three
largest values) of these features were computed for each image,
resulting in 30 features. For instance, field 3 is Mean Radius, field
13 is Radius SE, field 23 is Worst Radius.

All feature values are recoded with four significant digits.

Missing attribute values: none

Class distribution: 357 benign, 212 malignant

### Organizing the Data in Training and Test

To measure the performance of a classifier, it is essential to test the model on data it hasn't seen during training. To accomplish this, it's a common practice to partition your dataset into two distinct ones: a `training set` and a `test set`.
image.png
The training set is utilized for model development, training, and evaluation.

The test set is reserved for assessing how well the trained model generalizes to unseen data. This methodology provides valuable insights into the model's performance and its ability to handle new, unseen instances.

Thankfully, scikit-learn offers a convenient function called "train_test_split()" that simplifies the process of dividing your dataset into these distinct sets. You can import this function and then apply it to split your data as follows:

````python
label_names = data['target_names'] # clasification label names
labels = data['target'] # real labels
feature_names = data['feature_names']
features = data['data']
````

### 2. Code Breakdown

```python
train, test, train_labels, test_labels = train_test_split(features,
                                                          labels,
                                                          test_size=0.33,
                                                          random_state=42)
```

One function call producing **four** results at once, which is why there are four
names on the left. The order is fixed by the function:

| name | holds |
|---|---|
| `train` | the measurements the model learns from |
| `test` | the measurements it is judged on |
| `train_labels` | the answers for the training patients |
| `test_labels` | the answers for the test patients - kept hidden from the model |

- `random_state=42` - fixes the shuffle so everyone in this call gets the same
  split. Any number works; it just has to be the same number.

## Task 8 of 14 - Logistic Regression and convergence

### 1. Concept

**Logistic regression** draws a boundary through the 30-dimensional measurement
space and asks which side of it a tumour falls on. It is the standard first model
in clinical research: fast, stable on small datasets, and every coefficient can
be read and argued about.

It finds that boundary by improving a guess, over and over, until the improvement
stops mattering. That is **convergence**, and `max_iter` is the budget of
attempts it is allowed. Too small a budget and it stops while still mid-guess.

### 2. Code Breakdown

The three lines every scikit-learn model in this notebook uses:

```python
logreg = LogisticRegression(max_iter=1000)   # 1. create it, with settings
logreg.fit(train, train_labels)              # 2. learn from the training data
preds = logreg.predict(test)                 # 3. predict the unseen patients
```

Learn this shape - `create`, `.fit()`, `.predict()`. Naive Bayes, decision trees,
KNN and LightGBM all use exactly the same three steps, with a different name on
line one.

Then the score:

```python
accuracy_score(test_labels, preds)   # true answers first, predictions second
```

````python
from sklearn.linear_model import LogisticRegression

# TODO: Change max_iter from 1000 to 10
max_iter = 1000

logreg = LogisticRegression(max_iter=max_iter)
logreg.fit(train, train_labels)
logreg_preds = logreg.predict(test)

print(f"max_iter = {max_iter}")
print(f"Logistic regression accuracy: {accuracy_score(test_labels, logreg_preds):.4f}")
````

## Introduction to Deep model - Fully Connected Neural Networks (FCNN)

An FCNN consists of layers where every neuron in one layer is connected to every neuron in the next layer. This architecture is well-suited for structured or tabular data.

The script provided below builds, trains, evaluates, and uses a binary classification model using TensorFlow and Keras.

* **Model Architecture:** The network uses `keras.Sequential` to stack layers sequentially. It starts with a hidden layer of 64 neurons using the Rectified Linear Unit (`relu`) activation function, followed by a second hidden layer of 32 neurons. The final output layer uses a single neuron with a `sigmoid` activation function to output a probability between 0 and 1.
* **Compilation:** The model is compiled using the `adam` optimizer (an efficient gradient descent algorithm), `binary_crossentropy` as the loss function (standard for binary classification tasks), and tracks `accuracy` as a performance metric.
* **Training:** The `model.fit` function trains the network using your training features (`train`) and labels (`train_labels`) over 10 epochs with a batch size of 32.
* **Prediction and Evaluation:** The model generates predictions on the test set, thresholds the probability at 0.5 to convert it into binary classes (0 or 1), and evaluates performance using a confusion matrix, classification report, and overall accuracy score.

````python
# prompt: can you create a python code based on FCNN to do this prediction

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Define the model
model = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(train.shape[1],)),
    layers.Dense(32, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

# Compile the model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Train the model
model.fit(train, train_labels, epochs=10, batch_size=32)

# Make predictions
preds = (model.predict(test) > 0.5).astype(int)
preds = preds.flatten()

# Evaluate the model
print("Confusion Matrix:")
print(confusion_matrix(test_labels, preds))

print("\nClassification Report:")
print(classification_report(test_labels, preds))

# Accuracy
accuracy = accuracy_score(test_labels, preds)
print(f"\nAccuracy: {accuracy * 100:.2f}%")
````
